# Local Replot From Saved Attention-Matrix Data

Use this notebook to re-render cloud-side `plot_bundle.json` files locally and save edited figures into `drawing/attn_matrix/outputs/`.

```python
from pathlib import Path
import sys


def find_project_root(start: Path | None = None) -> Path:
    start = (start or Path.cwd()).resolve()
    for candidate in (start, *start.parents):
        if (candidate / 'drawing' / 'attn_matrix').is_dir() and (candidate / 'figs').is_dir():
            return candidate
    raise RuntimeError(f'Could not locate project root from {start}')


PAPER_ROOT = find_project_root()
ATTN_MATRIX_DIR = PAPER_ROOT / 'drawing' / 'attn_matrix'
if str(ATTN_MATRIX_DIR) not in sys.path:
    sys.path.insert(0, str(ATTN_MATRIX_DIR))

from replot_from_saved_data import (
    _load_bundle,
    infer_mode,
    replot_xparallel,
    replot_single_layer_flip,
)

ATTN_MATRIX_DIR
```

```python
# Edit these paths for the bundle you want to replot.
INPUT_BUNDLE = ATTN_MATRIX_DIR / "source/qwen3-4b/layer19/attn/layer_19_plot_bundle.json"
OUTPUT_IMAGE = ATTN_MATRIX_DIR / "outputs/qwen3-4b/layer19/attn_layer19_replot_notebook.png"
DPI = 220

INPUT_BUNDLE, OUTPUT_IMAGE
```

```python
summary_path, summary, bundle_path, npz = _load_bundle(str(INPUT_BUNDLE))
mode = infer_mode(npz, summary)
summary_path, bundle_path, mode
```

```python
OUTPUT_IMAGE.parent.mkdir(parents=True, exist_ok=True)
if mode == "xparallel":
    replot_xparallel(npz, summary, OUTPUT_IMAGE, DPI)
elif mode == "single_layer_flip":
    replot_single_layer_flip(npz, summary, OUTPUT_IMAGE, DPI)
else:
    raise ValueError(f"Unsupported mode: {mode}")

print(f"saved -> {OUTPUT_IMAGE}")
```
